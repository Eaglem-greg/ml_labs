```python
from linear_regression import  LinearRegression, Metrics
import numpy as np


x = np.linspace(-5, 5, 100).reshape(-1, 1)
y = 3 * x.ravel() + 5
```

```python
lg = LinearRegression()
```

```python
lg.fit(x, y)
```

```python
y_pred = lg.predict(x)
```

```python
lg.weights
```

```python
lg.bias
```

```python
metric = Metrics(y, y_pred)
```

```python
metric.mae()
```

```python
metric.mse()
```

```python
metric.r2()
```

```python
metric.report()
```
